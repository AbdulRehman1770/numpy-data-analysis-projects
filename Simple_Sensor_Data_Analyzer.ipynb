## Dataset

```python
import numpy as np

temperature = np.array([
    24.5, 25.1, 24.8, 25.6, 26.2, 25.9, 26.5, 27.1, 26.8, 27.4,
    27.0, 26.6, 25.8, 25.4, 24.9, 25.2, 25.7, 26.1, 26.9, 27.3,
    27.8, 28.1, 27.6, 26.9, 26.3, 25.8, 25.5, 25.0, 24.7, 24.9,
    25.3, 25.8, 26.4, 27.0, 27.5, 28.2, 28.7, 29.1, 29.5, 30.2,
    29.8, 29.3, 28.9, 28.4, 27.9, 27.2, 26.8, 26.1, 25.7, 25.3,
    24.8, 25.2, 25.9, 26.5, 27.1, 27.8, 28.4, 29.0, 29.6, 30.1,
    30.8, 31.2, 30.7, 30.0, 29.4, 28.8, 28.1, 27.5, 26.9, 26.3,
    25.8, 25.4, 25.0, 24.6, 24.9, 25.5, 26.2, 26.8, 27.4, 28.0,
    28.6, 29.2, 29.8, 30.5, 31.0, 31.8, 32.5, 33.2, 32.7, 31.9,
    31.2, 30.6, 29.9, 29.3, 28.7, 28.1, 27.6, 27.0, 26.5, 25.9
])
```

**Mean, Minimum, and Maximum Temperature.**

```python
mean = np.mean(temperature)
print("Mean Temperature Value: ", mean)

Min = np.min(temperature)
print("Minimum Temperature Value: ", Min)

Max = np.max(temperature)
print("Maximum Temperature Value: ", Max)
```

**Standard deviation**

```python
std = np.std(temperature)
print("Standard Deviation:", std)
```

**Identify abnormal readings**

```python
normal_reading = (temperature >= 15) & (temperature <= 25)

print("Normal Reading:", normal_reading)
print("Abnormal Reading:", ~normal_reading)
```

**Normalize**

```python
normalized = (temperature - Min) / (Max - Min)

print("Normalized Averages:", normalized)
```

**Normal / High / Low categories**

```python
normal_reading = (temperature >= 28) & (temperature <= 32)
Low_reading = (temperature >= 12) & (temperature <= 27)
High_reading = temperature > 32

print("Normal Reading:", temperature[normal_reading])
print("Low Reading:", temperature[Low_reading])
print("High Reading:", temperature[High_reading])
```

**Difference between consecutive readings**

```python
# Difference between consecutive readings

difference = np.diff(temperature)

print("Difference between consecutive readings:")
print(difference)
```

**Identify sudden temperature changes**

```python
sudden_change = np.abs(difference) >= 1

print("Sudden Temperature Changes:")
print(difference[sudden_change])
```

**Split dataset into training and testing sets**

```python
split_index = int(len(temperature) * 0.8)

training_data = temperature[:split_index]
testing_data = temperature[split_index:]

print("Training Data:")
print(training_data)

print("\nTesting Data:")
print(testing_data)
```
