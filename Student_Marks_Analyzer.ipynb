## Student Marks Analyzer

```python
# Using NumPy, analyze the marks of 10 students across 5 subjects. Perform the following tasks:

# Find the total marks of each student.
# Find the average marks of each student.
# Find the average marks of each subject.
# Find the highest and lowest marks in the entire dataset.
# Identify students who passed and failed. Passing marks = 50 in each subject.
# Rank the students based on their average marks.

# Assign a grade to each student:
# 80+ = A
# 70–79 = B
# 60–69 = C
# 50–59 = D
# Below 50 = Fail

# Find the student with the highest overall average.
# Normalize the students' average marks between 0 and 1.
```

## Dataset

```python
import numpy as np

students = np.array([
    [85, 78, 92, 88, 90],
    [72, 68, 75, 80, 70],
    [55, 62, 58, 65, 60],
    [91, 89, 95, 94, 90],
    [45, 52, 48, 50, 43],
    [76, 82, 79, 74, 85],
    [63, 59, 67, 61, 65],
    [88, 91, 84, 86, 89],
    [50, 47, 55, 52, 49],
    [69, 73, 71, 66, 75]
])
```

**Total marks of each student**

```python
total_marks = np.sum(students[:, 0:], axis=1)
print("Total Marks of Each Student: ", total_marks)
```

**Average Marks of Each Student**

```python
Average_Marks_Student = np.mean(students[:, 0:], axis=1)
print("Avergae Marks of Each Student :", Average_Marks_Student)
```

**Average Marks of Each Subject**

```python
Average_Marks_Subject = np.mean(students[:, 0:], axis=0)
print("Average Marks of Subject: ", Average_Marks_Subject)
```

**Highest and lowest Marks in Entire Dataset**

```python
Low_marks = np.min(students[:, :])
print("Low Marks", Low_marks)

High_marks = np.max(students[:, :])
print("High Marks", High_marks)
```

**Identify Passed and Failed Students**

```python
passed = np.all(students >= 50, axis=1)

print("Passed Students:", np.where(passed)[0] + 1)
print("Failed Students:", np.where(~passed)[0] + 1)
```

**Rank Students**

```python
rank = np.argsort(Average_Marks_Student)[::-1]

print("Ranking:")

for i, student in enumerate(rank):
    print("Rank", i + 1, ": Student", student + 1)
```

**Assign Grades**

```python
grades = np.where(Average_Marks_Student >= 80, "A",
         np.where(Average_Marks_Student >= 70, "B",
         np.where(Average_Marks_Student >= 60, "C",
         np.where(Average_Marks_Student >= 50, "D", "Fail"))))

print("Grades:", grades)
```

**Student With Highest Average**

```python
Highest_Average = np.max(Average_Marks_Student)

Student_name = np.where(Average_Marks_Student == Highest_Average)

print("Highest Average:", Highest_Average)
print("Student:", Student_name[0] + 1)
```

**Normalize Average Marks**

```python
normalized = (Average_Marks_Student - np.min(Average_Marks_Student)) / (np.max(Average_Marks_Student) - np.min(Average_Marks_Student))

print("Normalized Averages:", normalized)
```

**Split dataset into training and testing sets**

```python
split_index = int(len(students) * 0.7)

training_data = students[:split_index]
testing_data = students[split_index:]

print("Training Data:")
print(training_data)

print("\nTesting Data:")
print(testing_data)
```
